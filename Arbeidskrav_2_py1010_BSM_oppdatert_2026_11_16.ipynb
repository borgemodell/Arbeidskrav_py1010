```python
"""

Arbeidskrav 2 py1010 
Børge Modell
2026 11 16

Oppgave 5

"""
def fun_areal_og_omkrets (a, b):
    A = (np.pi*a**2)/2 + (a*b)/2
    c = math.hypot(a, b)
    O = np.pi*a + (b + c)
    return (A, O)

radius = 5
katet = 10

(areal, omkrets) = fun_areal_og_omkrets(radius, katet)
print("Arealet på figuren er", areal)
print("Den ytre omkretsen på figuren er", omkrets)
```
